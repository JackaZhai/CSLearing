```python
import math
import matplotlib.pyplot as plt

x_start = 0
print(math.sin(x_start))
x_end = 2 * math.pi
print(math.sin(x_end))

num = 37
step = (x_end - x_start) / (num-1)
x_array = [x_start + i * step for i in range(num)]
zero_array = [0 for i in range(num)]

plt.plot(x_array, zero_array,marker = '.',markersize = 8, markerfacecolor = 'w',markeredgecolor = 'k')
plt.text(x_start, 0, '0')
plt.text(x_end, 0, '2π')
plt.axis('off')
plt.show()

y_array = [math.sin(x) for x in x_array]
plt.plot(x_array, y_array, marker='.', markersize=8, markerfacecolor='w', markeredgecolor='k')
plt.text(x_start, -0.1, '0')
plt.text(x_end, 0.1, '2π')
plt.axis('off')
plt.axhline(y=0, color = 'k', linestyle='--', linewidth=0.25)
plt.show()
```

```python
import random 
import statistics
import matplotlib.pyplot as plt

num_flips = 1000

results = []
running_means = []

for i in range(num_flips):
    flip = random.choice([0.4, 1])
    results.append(flip)
    running_mean = statistics.mean(results)
    running_means.append(running_mean)

plt.plot(running_means, marker='.', markersize=2, markerfacecolor='w', markeredgecolor='k')
plt.axhline(y=0.5, color='k', linestyle='--', linewidth=0.25)
```
