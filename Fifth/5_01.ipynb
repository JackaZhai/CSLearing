# Python中的append方法

`append()`是Python列表(list)对象的一个方法，用于在列表末尾添加一个新元素。

### 语法
```python
list.append(element)
```

### 特点
- 只能添加一个元素到列表末尾
- 直接修改原列表，不返回新列表
- 可以添加任何类型的对象（数字、字符串、列表、字典等）

````python
# 演示append方法的使用

# 创建一个空列表
fruits = []
print(f"初始列表: {fruits}")

# 使用append添加元素
fruits.append("苹果")
print(f"添加一个元素后: {fruits}")

# 继续添加更多元素
fruits.append("香蕉")
fruits.append("橙子")
print(f"添加多个元素后: {fruits}")

# append可以添加任何类型的对象
fruits.append(100)  # 添加数字
fruits.append(["草莓", "蓝莓"])  # 添加列表
fruits.append({"name": "芒果", "color": "黄色"})  # 添加字典
print(f"添加不同类型元素后: {fruits}")

# 注意：append只能一次添加一个元素
print("\n区别：使用append vs 使用extend/+")
list1 = [1, 2, 3]
list2 = [1, 2, 3]
list1.append([4, 5])  # append会将整个列表作为一个元素添加
list2.extend([4, 5])  # extend会将列表中的每个元素分别添加
print(f"使用append: {list1}")
print(f"使用extend: {list2}")
````
