### 图像存储方式:RGB
使用[255][255][3]数组存储

(0,0,0)黑色
(255,255,255)白色
(255,0,0)红色
(0,255,0)绿色

```python
import cv2
import os
import numpy as np

blue = np.array([0, 0, 255], dtype=np.uint8) # BGR
green = np.array([0, 255, 0], dtype=np.uint8) # BGR
red = np.array([255, 0, 0], dtype=np.uint8) # BGR 写错了

#保存红绿蓝色的图片
img = np.zeros((100, 100, 3), dtype=np.uint8) # 创建一个100*100的黑色图片
img[:] = red 
cv2.imwrite('blue.jpg', img) # 保存图片

imread = cv2.imread('blue.jpg', cv2.IMREAD_UNCHANGED)
'''
imread ('path', flags) -> retval
'''

cv2.imshow('pic', imread)
cv2.waitKey(0)
```
