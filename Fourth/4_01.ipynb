```python
import numpy as np

x_array = np.arange(5.0,10,0.1)# Create a 1D array with values from 0 to 9

def functionplus(x):
    return x + 1

print(functionplus(x_array)) # Call the function with the array
```

```python
import numpy as np
a = np.array([[1, 2, 3], [4, 5, 6]]) # Create a 2D array
b = np.array([[7, 8, 9], [10, 11, 12]]) # Create another 2D array
c = np.concatenate((a, b), axis=0) # Concatenate along the first axis
print(c) # Print the concatenated array
```

```python
import copy
a_2_layers = [1,2,[3,4,5]] # Create a list with 2 layers
d_2_layers = a_2_layers.copy() # Create a shallow copy of the list
e_2_layers = copy.deepcopy(a_2_layers) # Create a deep copy of the list

print(a_2_layers is d_2_layers)
print(a_2_layers[2] is d_2_layers[2])

print(a_2_layers is e_2_layers)
print(a_2_layers[2] is e_2_layers[2])
```
